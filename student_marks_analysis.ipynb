```python
import matplotlib.pyplot as plt

students = []

print("================================")
print("   STUDENT MARKS ANALYSIS SYSTEM")
print("================================")

n = int(input("Enter number of students: "))

subjects = ["Science", "Maths", "English"]

# Student Details
for i in range(n):

    print("\nEnter details of Student", i + 1)

    name = input("Enter student name: ")
    roll = input("Enter roll number: ")

    marks = []

    for subject in subjects:
        while True:
            mark = float(input("Enter " + subject + " marks: "))

            if 0 <= mark <= 100:
                marks.append(mark)
                break
            else:
                print("Please enter marks between 0 and 100")

    total = sum(marks)
    percentage = total / 3

    # Grade Calculation
    if percentage >= 90:
        grade = "A+"
    elif percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    else:
        grade = "Fail"

    students.append({
        "name": name,
        "roll": roll,
        "marks": marks,
        "total": total,
        "percentage": percentage,
        "grade": grade
    })


# Display Student Results
print("\n========== STUDENT RESULTS ==========")

for s in students:
    print("\nName:", s["name"])
    print("Roll Number:", s["roll"])
    print("Total Marks:", s["total"], "/300")
    print("Percentage:", round(s["percentage"], 2), "%")
    print("Grade:", s["grade"])


# Find Topper
topper = max(students, key=lambda x: x["percentage"])

print("\n========== CLASS ANALYSIS ==========")
print("Class Topper:", topper["name"])
print("Topper Percentage:", round(topper["percentage"], 2), "%")

# Class Average
class_average = sum(s["percentage"] for s in students) / n
print("Class Average:", round(class_average, 2), "%")

# Subject-wise Average
averages = []

print("\nSubject-wise Average Marks:")

for i, subject in enumerate(subjects):
    avg = sum(s["marks"][i] for s in students) / n
    averages.append(avg)
    print(subject, ":", round(avg, 2))

# Student Performance Graph
names = [s["name"] for s in students]
percentages = [s["percentage"] for s in students]

plt.figure(figsize=(8, 5))
plt.bar(names, percentages)
plt.xlabel("Student Names")
plt.ylabel("Percentage")
plt.title("Student Performance Analysis")
plt.ylim(0, 100)
plt.show()

# Subject Average Graph
plt.figure(figsize=(8, 5))
plt.bar(subjects, averages)
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.title("Subject Wise Average Marks")
plt.ylim(0, 100)
plt.show()

print("\nProject Completed Successfully!")
```
